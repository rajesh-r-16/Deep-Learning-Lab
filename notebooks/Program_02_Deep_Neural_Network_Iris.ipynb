```python
# Import the package or class needed in this experiment.
import numpy as np
# Import the package or class needed in this experiment.
import pandas as pd
# Import the package or class needed in this experiment.
import tensorflow as tf
# Import the package or class needed in this experiment.
import matplotlib.pyplot as plt
# Import the package or class needed in this experiment.
from pathlib import Path
# Import the package or class needed in this experiment.
from sklearn.datasets import load_iris
# Import the package or class needed in this experiment.
from sklearn.model_selection import train_test_split
# Import the package or class needed in this experiment.
from sklearn.preprocessing import StandardScaler
# Import the package or class needed in this experiment.
from sklearn.metrics import classification_report, confusion_matrix, ConfusionMatrixDisplay
# Import the package or class needed in this experiment.
from tensorflow.keras.models import Sequential
# Import the package or class needed in this experiment.
from tensorflow.keras.layers import Dense, Dropout

# Perform this processing step in the experiment.
tf.keras.utils.set_random_seed(42)
# Create or define the folder used to store program outputs.
OUT = Path("/content/DL_Lab_Outputs/Program_2")
# Create or define the folder used to store program outputs.
OUT.mkdir(parents=True, exist_ok=True)

# Assign the resulting value to `iris`.
iris = load_iris()
# Assign the resulting value to `X, y`.
X, y = iris.data, iris.target
# Assign the resulting value to `X_train, X_test, y_train, y_test`.
X_train, X_test, y_train, y_test = train_test_split(
    # Assign the resulting value to `X, y, test_size`.
    X, y, test_size=0.2, random_state=42, stratify=y
# Perform this processing step in the experiment.
)

# Assign the resulting value to `scaler`.
scaler = StandardScaler()
# Assign the resulting value to `X_train`.
X_train = scaler.fit_transform(X_train)
# Assign the resulting value to `X_test`.
X_test = scaler.transform(X_test)

# Assign the resulting value to `model`.
model = Sequential([
    # Assign the resulting value to `Dense(16, activation`.
    Dense(16, activation="relu", input_shape=(4,)),
    # Assign the resulting value to `Dense(12, activation`.
    Dense(12, activation="relu"),
    # Assign the resulting value to `Dense(8, activation`.
    Dense(8, activation="relu"),
    # Assign the resulting value to `Dense(3, activation`.
    Dense(3, activation="softmax")
# Perform this processing step in the experiment.
])
# Configure the optimizer, loss function, and evaluation metrics.
model.compile(optimizer="adam", loss="sparse_categorical_crossentropy", metrics=["accuracy"])
# Perform this processing step in the experiment.
model.summary()

# Train the model using the prepared training data.
history = model.fit(X_train, y_train, epochs=50, batch_size=8,
                    # Assign the resulting value to `validation_split`.
                    validation_split=0.2, verbose=1)
# Evaluate the trained model on the test dataset.
loss, accuracy = model.evaluate(X_test, y_test, verbose=0)
# Generate predictions for the supplied input samples.
probabilities = model.predict(X_test, verbose=0)
# Assign the resulting value to `predicted`.
predicted = np.argmax(probabilities, axis=1)

# Display the current result or evaluation metric.
print(f"Test Accuracy: {accuracy*100:.2f}%")
# Display the current result or evaluation metric.
print("\nClassification Report:")
# Display the current result or evaluation metric.
print(classification_report(y_test, predicted, target_names=iris.target_names))
# Display the current result or evaluation metric.
print("Predicted Classes:", predicted)
# Display the current result or evaluation metric.
print("Actual Classes   :", y_test)

# Save experiment results to an output file.
pd.DataFrame({"actual": y_test, "predicted": predicted}).to_csv(OUT/"iris_predictions.csv", index=False)
# Save experiment results to an output file.
pd.DataFrame(history.history).to_csv(OUT/"training_history.csv", index=False)
# Assign the resulting value to `ConfusionMatrixDisplay.from_predictions(y_test, predicted, display_labels`.
ConfusionMatrixDisplay.from_predictions(y_test, predicted, display_labels=iris.target_names)
# Create, format, save, or display the visualization.
plt.title("Iris DNN Confusion Matrix")
# Create, format, save, or display the visualization.
plt.tight_layout()
# Create, format, save, or display the visualization.
plt.savefig(OUT/"confusion_matrix.png", dpi=150)
# Create, format, save, or display the visualization.
plt.show()
```
