```python
# Install required package
# Install or run the required command in the Colab runtime.
!pip -q install gensim

# Import the package or class needed in this experiment.
from gensim.models import Word2Vec
# Import the package or class needed in this experiment.
import numpy as np
# Import the package or class needed in this experiment.
import pandas as pd
# Import the package or class needed in this experiment.
from pathlib import Path

# Create or define the folder used to store program outputs.
OUT = Path("/content/DL_Lab_Outputs/Program_1")
# Create or define the folder used to store program outputs.
OUT.mkdir(parents=True, exist_ok=True)

# Sample document corpus (tokenized sentences)
# Assign the resulting value to `corpus`.
corpus = [
    # Provide the next argument or item in this multi-line expression.
    ["machine", "learning", "is", "fun"],
    # Provide the next argument or item in this multi-line expression.
    ["deep", "learning", "uses", "neural", "networks"],
    # Provide the next argument or item in this multi-line expression.
    ["deep", "learning", "is", "powerful"],
    # Provide the next argument or item in this multi-line expression.
    ["machine", "learning", "uses", "data"],
    # Provide the next argument or item in this multi-line expression.
    ["neural", "networks", "learn", "patterns"]
# Perform this processing step in the experiment.
]

# Train Word2Vec model
# Assign the resulting value to `model`.
model = Word2Vec(
    # Assign the resulting value to `sentences`.
    sentences=corpus,
    # Assign the resulting value to `vector_size`.
    vector_size=50,
    # Assign the resulting value to `window`.
    window=2,
    # Assign the resulting value to `min_count`.
    min_count=1,
    # Assign the resulting value to `workers`.
    workers=1,
    # Assign the resulting value to `seed`.
    seed=42,
    # Assign the resulting value to `epochs`.
    epochs=100
# Perform this processing step in the experiment.
)

# Assign the resulting value to `word`.
word = "learning"
# Display the current result or evaluation metric.
print(f"Embedding for '{word}':")
# Display the current result or evaluation metric.
print(model.wv[word])
# Display the current result or evaluation metric.
print("\nWords similar to 'learning':")
# Display the current result or evaluation metric.
print(model.wv.most_similar(word, topn=5))

# Save all word vectors and similar-word results
# Assign the resulting value to `vectors`.
vectors = pd.DataFrame(model.wv.vectors, index=model.wv.index_to_key)
# Assign the resulting value to `vectors.index.name`.
vectors.index.name = "word"
# Save experiment results to an output file.
vectors.to_csv(OUT/"word_embeddings.csv")
# Create or transform the data used by the experiment.
pd.DataFrame(model.wv.most_similar(word, topn=5),
             # Save experiment results to an output file.
             columns=["word", "similarity"]).to_csv(OUT/"similar_words.csv", index=False)
# Save the trained model or reusable objects.
model.save(str(OUT/"word2vec.model"))
# Display the current result or evaluation metric.
print("\nSaved word_embeddings.csv, similar_words.csv and word2vec.model")
```
