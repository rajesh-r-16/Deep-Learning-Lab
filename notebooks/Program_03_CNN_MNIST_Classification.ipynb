```python
# Import the package or class needed in this experiment.
import numpy as np
# Import the package or class needed in this experiment.
import pandas as pd
# Import the package or class needed in this experiment.
import tensorflow as tf
# Import the package or class needed in this experiment.
import matplotlib.pyplot as plt
# Import the package or class needed in this experiment.
from pathlib import Path
# Import the package or class needed in this experiment.
from tensorflow.keras.datasets import mnist
# Import the package or class needed in this experiment.
from tensorflow.keras.utils import to_categorical
# Import the package or class needed in this experiment.
from tensorflow.keras.models import Sequential
# Import the package or class needed in this experiment.
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout

# Perform this processing step in the experiment.
tf.keras.utils.set_random_seed(42)
# Create or define the folder used to store program outputs.
OUT = Path("/content/DL_Lab_Outputs/Program_3")
# Create or define the folder used to store program outputs.
OUT.mkdir(parents=True, exist_ok=True)

# Assign the resulting value to `(x_train, y_train), (x_test, y_test)`.
(x_train, y_train), (x_test, y_test) = mnist.load_data()
# Assign the resulting value to `x_train`.
x_train = x_train.reshape((-1, 28, 28, 1)).astype("float32") / 255.0
# Assign the resulting value to `x_test`.
x_test = x_test.reshape((-1, 28, 28, 1)).astype("float32") / 255.0
# Assign the resulting value to `y_train_cat`.
y_train_cat = to_categorical(y_train, 10)
# Assign the resulting value to `y_test_cat`.
y_test_cat = to_categorical(y_test, 10)

# Assign the resulting value to `model`.
model = Sequential([
    # Assign the resulting value to `Conv2D(32, (3,3), activation`.
    Conv2D(32, (3,3), activation="relu", input_shape=(28,28,1)),
    # Perform this processing step in the experiment.
    MaxPooling2D((2,2)),
    # Assign the resulting value to `Conv2D(64, (3,3), activation`.
    Conv2D(64, (3,3), activation="relu"),
    # Perform this processing step in the experiment.
    MaxPooling2D((2,2)),
    # Perform this processing step in the experiment.
    Flatten(),
    # Assign the resulting value to `Dense(128, activation`.
    Dense(128, activation="relu"),
    # Perform this processing step in the experiment.
    Dropout(0.5),
    # Assign the resulting value to `Dense(10, activation`.
    Dense(10, activation="softmax")
# Perform this processing step in the experiment.
])
# Configure the optimizer, loss function, and evaluation metrics.
model.compile(optimizer="adam", loss="categorical_crossentropy", metrics=["accuracy"])
# Train the model using the prepared training data.
history = model.fit(x_train, y_train_cat, epochs=5, batch_size=64,
                    # Assign the resulting value to `validation_split`.
                    validation_split=0.1, verbose=1)
# Evaluate the trained model on the test dataset.
loss, accuracy = model.evaluate(x_test, y_test_cat, verbose=0)
# Generate predictions for the supplied input samples.
predictions = model.predict(x_test[:10], verbose=0)
# Assign the resulting value to `predicted_classes`.
predicted_classes = np.argmax(predictions, axis=1)

# Display the current result or evaluation metric.
print(f"Test Accuracy: {accuracy*100:.2f}%")
# Display the current result or evaluation metric.
print("Predicted classes:", predicted_classes)
# Display the current result or evaluation metric.
print("Actual classes:   ", y_test[:10])

# Save experiment results to an output file.
pd.DataFrame({"actual": y_test[:10], "predicted": predicted_classes}).to_csv(OUT/"mnist_predictions.csv", index=False)
# Save experiment results to an output file.
pd.DataFrame(history.history).to_csv(OUT/"training_history.csv", index=False)
# Create, format, save, or display the visualization.
plt.plot(history.history["accuracy"], label="Training accuracy")
# Create, format, save, or display the visualization.
plt.plot(history.history["val_accuracy"], label="Validation accuracy")
# Create, format, save, or display the visualization.
plt.xlabel("Epoch"); plt.ylabel("Accuracy"); plt.title("MNIST CNN Accuracy")
# Create, format, save, or display the visualization.
plt.legend(); plt.tight_layout()
# Create, format, save, or display the visualization.
plt.savefig(OUT/"accuracy_curve.png", dpi=150); plt.show()
# Save the trained model or reusable objects.
model.save(OUT/"mnist_cnn.keras")
```
